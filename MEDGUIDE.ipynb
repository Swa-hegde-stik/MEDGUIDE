# Building Recommendation System

```python
symptoms=pd.read_csv('/content/drive/MyDrive/Projects/Medicine_recommender/symtoms_df.csv')
```

```python
precautions=pd.read_csv('/content/drive/MyDrive/Projects/Medicine_recommender/precautions_df.csv')
```

```python
workouts=pd.read_csv('/content/drive/MyDrive/Projects/Medicine_recommender/workout_df.csv')
medications=pd.read_csv('/content/drive/MyDrive/Projects/Medicine_recommender/medications.csv')
description=pd.read_csv('/content/drive/MyDrive/Projects/Medicine_recommender/description.csv')
diets=pd.read_csv('/content/drive/MyDrive/Projects/Medicine_recommender/diets.csv')
```

```python
symp_dict=dict(zip(le.classes_,le.transform(le.classes_)))
```

```python
svc = SVC()
svc.fit(x_train, y_train)

def predict(symptoms_seen):
    input_vector = np.zeros(len(x.columns))

    for symptom in symptoms_seen:
        if symptom in x.columns:
            input_vector[x.columns.get_loc(symptom)] = 1

    prediction = svc.predict([input_vector])[0]

    return le.inverse_transform([prediction])[0]
```

```python
symptoms_input=input("provide the symptoms seperated by comma:")
user_symptoms=[]
for symptom in symptoms_input.split(','):
  user_symptoms.append(symptom.strip())

predicted_disease=predict(user_symptoms)
print(predicted_disease)


```

```python
print(description.head())
print()
print(precautions.head())
print()
print(workouts.head())
print()
print(medications.head())
print()
print(diets.head())
```

```python
#Function for all recommendations
def get_recommendations(disease):

    result = {}

    # Description
    result["description"] = description[
        description["Disease"] == disease
    ]["Description"].values

    # Precautions
    result["precautions"] = precautions[
        precautions["Disease"] == disease
    ].values

    # Workout
    result["workout"] = workouts[
        workouts["disease"] == disease
    ]["workout"].values

    # Medication
    result["medications"] = medications[
        medications["Disease"] == disease
    ]["Medication"].values

    # Diet
    result["diet"] = diets[
        diets["Disease"] == disease
    ]["Diet"].values

    return result
```

```python
recommendations = get_recommendations(predicted_disease)

print("Disease:", predicted_disease)
print("\nDescription:", recommendations["description"])
print("\nPrecautions:", recommendations["precautions"])
print("\nWorkout:", recommendations["workout"])
print("\nMedications:", recommendations["medications"])
print("\nDiet:", recommendations["diet"])
```

# Combining both Function

```python
def Health_Recommendation():
  Symptoms_Seen=input("provide the symptoms seperated by comma:")

  #convert input to list
  user_symptoms=[]
  for symptoms in Symptoms_Seen.split(','):
    user_symptoms.append(symptoms.strip().lower())

  #creating input vector for the model
  input_vector=np.zeros(len(x.columns))
  valid_symptoms=[]

  for symptoms in user_symptoms:
    input_vector[x.columns.get_loc(symptoms)]=1
    valid_symptoms.append(symptoms)

  #checking whether symptoms are valid
  if len(valid_symptoms)==0:
    print('symptoms entered are invalid')
    return

  #Predicting disease
  prediction=svc.predict([input_vector])[0]
  predicted_disease=le.inverse_transform([prediction])[0]

  #Description
  disease_description=description[description['Disease']==predicted_disease]['Description'].iloc[0]

  #Precautions
  precautions_row=precautions[precautions['Disease']==predicted_disease]
  disease_precautions=precautions_row[
      ["Precaution_1", "Precaution_2",
         "Precaution_3", "Precaution_4"]
  ].iloc[0].tolist()

  #Workouts
  workout_list=workouts[workouts["disease"] == predicted_disease]["workout"].tolist()

  #Medication
  medication = medications[medications["Disease"] == predicted_disease]["Medication"].iloc[0]

  #Diets
  diet = diets[diets["Disease"] == predicted_disease]["Diet"].iloc[0]

  #Displaying everything

  print("\n" + "*"*50)
  print("             HEALTH RECOMMENDATION")
  print("*"*50)

  print("\nSymptoms entered:")
  print(", ".join(valid_symptoms))

  print("\nPredicted Disease:")
  print(predicted_disease)

  print("\nDescription:")
  print(disease_description)

  print("\nPrecautions:")
  for p in disease_precautions:
     print("-", p)

  print("\nWorkout:")
  for w in workout_list:
    print("-", w)

  import ast
  medication_list=ast.literal_eval(medication)
  diet_list=ast.literal_eval(diet)

  print("\nDiet:")
  for d in diet_list:
    print('-',d)

  print("\nMedication Information:")
  for m in medication_list:
    print('-',m)

  print("\n" + "*"*50)
```

```python
Health_Recommendation()
```
